```python
#print total-marks=[35,36,37,38,39,40]-----while loop
marks=[35,36,37,38,39,40]

total_marks=0

i=0
while i<=5:
    total_marks=total_marks+marks[i]
    
    i=i+1
print(total_marks)    
```
