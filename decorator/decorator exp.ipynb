```python
def testcase(func):
    def inner(a,b):
        if b==0:
            print("Can't Divide by Zero")
        else:
            return func(a,b)
    return inner        
@testcase
def calc(a,b):
    print(a/b)
calc(10,2)   #5.0
calc(10,0)    #ZeroDivisionError: division by zero
```
